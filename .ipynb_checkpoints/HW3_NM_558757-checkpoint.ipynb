Given the following function,
$$f(x) = sin(x)$$
and it's derivative,
$$f(x) = cos(x)$$
I will approximate it's derivative at $f(3)$, calculate it's true error with and without the richardson extraploation formula,
$$D_{richardson} = \frac{4D(h/2) - D(h)}{p^2 - 1}$$
for $p = 2$ and D as the central difference
$$f'(x) = D_{plain} = \frac{f(x + h) - f(x - h)}{2h}$$
and compare the results

## SOURCE CODE

```python
import numpy as np
import matplotlib.pyplot as plt

p = 2

def f(x):
    return np.sin(x)

def f_prime(x):
    return np.cos(x)

def D(x, h):
    return (f(x+h) - f(x-h)) / (2*h)
    
def richardson_extrapolation(x, h):
    return (4*D(x, h/2) - D(x, h))/((p**2)-1)

def relative_error(x, h, function):
    return abs((f_prime(x) - function(x, h))/f_prime(x))

x = 3
h = np.logspace(0, -16, 500)
h_size = np.size(h)

error_plain = relative_error(x, h, D)
error_rich = relative_error(x, h ,richardson_extrapolation)

plt.figure(figsize=(12,9))
plt.title('True Error Estimation')
plt.plot(h, error_plain, label='Plain')
plt.plot(h, error_rich, label='Richardson Extrapolation')
plt.grid()
plt.legend()
plt.xlabel('h')
plt.ylabel('Error')
plt.xscale('log')
plt.yscale('log')
plt.gca().invert_xaxis()
```

## Conclusion
As you can observe from the graph, using the richardson extrapolation, the true error decreases much faster and reaches a much lower bound compared without it.
